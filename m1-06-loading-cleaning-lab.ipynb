```python
import pandas as pd
```

```python
tickets = pd.read_csv("data_safe_copy.csv",index_col=0)
tickets.head()
```

```python
tickets.info()
```

```python
#t2 means task 2
tickets_t2 = tickets.copy()
tickets_t2['opened_at'] = pd.to_datetime(tickets_t2['opened_at'])
tickets_t2.head()
```

```python
tickets_t2.info()
```

```python
tickets_t2['category'] = tickets_t2['category'].str.lower().str.strip()
tickets_t2['category'].unique()
```

```python
tickets_t2['priority'] = tickets_t2['priority'].str.lower().str.strip()
tickets_t2['priority'].unique()
```

```python
#task3
tickets_t3 = tickets_t2.copy()
na_values = "unknown"
tickets_t3['resolution_minutes'] = pd.to_numeric(tickets_t3['resolution_minutes'], errors='coerce')
print(tickets_t3.isnull().sum())
print(tickets['resolution_minutes'].isnull().sum())
print((tickets['resolution_minutes'] == 'unknown').sum())
tickets_t3
```

```python
#task4
tickets_clean = tickets_t3.copy()
tickets_clean.dropna(subset=['resolution_minutes'], inplace=True)
print(tickets_clean.isnull().sum())
```

```python
#task 4 verification
print(tickets_t3.shape[0])
print(tickets_clean.shape[0])
print((tickets_clean['resolution_minutes'] < 0).any())
print("Unique Categories:")
print(tickets_clean['category'].unique())

print("\nUnique Priorities:")
print(tickets_clean['priority'].unique())
```

```python
#task5
total = len(tickets)
clean = len(tickets_clean)
missing = tickets['resolution_minutes'].isna().sum()+(tickets['resolution_minutes'] == 'unknown').sum()
avg_res = tickets_clean['resolution_minutes'].mean()
max_res = tickets_clean['resolution_minutes'].max()

summary = {
    'total_records': total,
    'clean_records': clean,
    'missing_resolution': missing,
    'avg_resolution': avg_res,
    'max_resolution': max_res
}

print("Summary Metrics:")
print(summary)

verification_check = (summary['clean_records'] + summary['missing_resolution']) == summary['total_records']

print("\nVerification Passed:", verification_check)
```
